# 1. Créer des tableaux NumPy

NumPy permet de créer des tableaux numériques pour stocker, organiser
et manipuler efficacement des données.

Un tableau NumPy (`ndarray`) peut être :
- unidimensionnel : une suite de valeurs ;
- bidimensionnel : des lignes et des colonnes ;
- multidimensionnel : plusieurs axes de données.

Dans cette section, nous allons apprendre à créer des tableaux
à partir de valeurs existantes, à générer des tableaux initialisés,
à créer des séquences et à générer des données aléatoires.

```python
# Importer NumPy sous son alias conventionnel np
import numpy as np
```

## 1.1 Créer un tableau avec np.array()

`np.array()` construit un tableau NumPy à partir de valeurs existantes.

Syntaxe générale :

    np.array(valeurs)

Cette fonction est utile lorsque nous disposons déjà de données,
par exemple des montants de ventes, des températures ou des quantités.

```python
# Montants de quatre ventes, exprimés en FCFA
montants = np.array([1000, 2500, 3000, 4500])
print(montants)
```

```python
# verifie le type de l'object ou la nature 
print(type(montants))
# verifie le type de des elements du tableau
print(montants.dtype)
```

## 1.2 — Créer des tableaux initialisés : np.zeros() et np.ones()

NumPy propose notamment :
- np.zeros() : crée un tableau rempli de zéros.
- np.ones() : crée un tableau rempli de uns.
Ces fonctions sont utiles pour initialiser des tableaux avant d'effectuer des calculs numériques.

```python
# crée un tab de 5elements init à zéro
ventes = np.zeros(5)
print(ventes)

# Par défaut, np.zeros() un tableau de nombres décimaux de type float
ventes_int = np.zeros(5, dtype=int)
print(ventes_int)
print(ventes_int.dtype)

# cree un tab de 4 elements init à 1
coef = np.ones(4, dtype=int)
print(coef)

# on peut aussi crée un tab de 2 dimensions
matrice = np.ones((2, 3), dtype=int)
print(matrice)

# ce que je peux retenir : np.zeros(5) crée cinq éléments sur un seul axe, 
# tandis que np.zeros((2, 3)) crée deux lignes et trois colonnes.
```

## 1.3 — Créer des séquences avec np.arange()

np.arange() génère une suite de valeurs espacées par un pas défini.

Syntaxe : 
np.arange(debut, fin, pas)

- debut : première valeur incluse.
- fin : limite exclue.
- pas : écart entre deux valeurs successives.

C'est particulièrement pratique pour générer des indices, des séquences numériques ou des séries de valeurs.

```python
sequence = np.arange(5)
sequence
# Si seul un argument est fourni, NumPy commence à zéro.
```

```python
# Générer les valeurs de 2 à 10 exclu, avec un pas de 2
nombres = np.arange(2, 10, 2)
print(nombres)
```

```python
# Générer une séquence de 0 à 1 exclu, par pas de 0.2
decimaux = np.arange(0, 1, 0.2)
print(decimaux)
```

## 1.4 — Générer un nombre précis de points avec np.linspace()
Cellule Markdown
Contrairement à np.arange(), qui définit principalement un pas, np.linspace() définit un nombre de points à répartir entre deux bornes.

Syntaxe : np.linspace(debut, fin, nombre_de_points)

Par défaut, la valeur de fin est incluse.


```python
points = np.linspace(0, 10, 5)
print(points)
```

## 1.5 — Générer des données aléatoires avec np.random
En Data Engineering et en analyse de données, on utilise parfois des données simulées pour tester un traitement avant de disposer de données réelles.

Le module np.random permet de générer des valeurs aléatoires. Nous allons utiliser trois opérations :
- np.random.default_rng() : créer un générateur aléatoire.
- rng.integers() : générer des entiers.
- rng.random() : générer des nombres décimaux entre 0 inclus et 1 exclu.

```python
# On crée le générateur aléatoire avec une graine fixe
rng = np.random.default_rng(42)

# generez des entiers 
valeurs = rng.integers(1, 100, size=10)
print(valeurs)
```

```python
# générez des nombres decimmaux 
nbres_decimaux = rng.random(5)
print(nbres_decimaux)
```

# 2. Comprendre la structure des tableaux

Un tableau NumPy possède plusieurs propriétés essentielles :

- `ndim` : nombre de dimensions ;
- `shape` : taille de chaque dimension ;
- `size` : nombre total d'éléments ;
- `dtype` : type des éléments stockés.

Ces propriétés permettent de vérifier la structure des données
avant d'effectuer des calculs ou de les transmettre à un autre traitement.

```python
# on crée un tab de 4 montants
montants1 = np.array([1000, 2900, 3200, 5000])

# nbre de dimensions
print("Dimension :", montants1.ndim)
# nbre d'elements sur chaque axe
print("Forme :", montants1.shape)
# nbre total d'elements
print("Taille :", montants1.size)
# type d'elements
print("Type :", montants1.dtype)
```

```python
# tab 2 dimensions 
# 2 lignes represente 2 jours de ventes

ventes1 = np.array([
    [1000, 2000, 3000], 
    [4000, 5000, 6000]
])
print("Dimension :", ventes1.ndim)
print("Forme :", ventes1.shape)
print("Taille :", ventes1.size)
print("Type :", ventes1.dtype)
```

# 3. Opérations vectorisées

La vectorisation consiste à appliquer une opération à un tableau
entier sans écrire explicitement une boucle Python.

Elle rend le code plus concis et permet généralement de profiter
des optimisations numériques de NumPy.

Nous allons tester :
- l'addition et la soustraction ;
- la multiplication et la division ;
- les opérations entre deux tableaux ;
- les comparaisons élément par élément.

```python
# Prix de quatre produits en FCFA
prix1 = np.array([1000, 2000, 3000, 4000])

# Augmenter chaque prix1 de 500 FCFA
prix1_augmentes = prix1 + 500

# Réduire chaque prix1 de 10 %
prix1_reduits = prix1 * 0.90

# Doubler chaque prix1
prix1_doubles = prix1 * 2

print("prix1 initiaux :", prix1)
print("prix1 augmentés :", prix1_augmentes)
print("prix1 réduits :", prix1_reduits)
print("prix1 doublés :", prix1_doubles)
```

```python
# Opérations entre deux tableaux
# Quantités vendues pour chaque produit
quantites = np.array([2, 3, 1, 5])

# Calculer le chiffre d'affaires par produit
chiffre_affaires = prix1 * quantites

print(chiffre_affaires)
```

```python
# Comparaisons
#identifier les prix >= 3000
masque_prix = prix1 >= 3000
print(masque_prix)
```

# 4. Broadcasting

Le broadcasting est le mécanisme qui permet à NumPy d'effectuer
certaines opérations entre des tableaux de formes différentes.

Lorsqu'une opération compare les dimensions de deux tableaux,
NumPy examine leurs formes de droite à gauche.

Deux dimensions sont compatibles si :
- elles sont égales ;
- ou si l'une d'elles vaut 1.

Si toutes les dimensions sont compatibles, NumPy peut étendre
virtuellement les valeurs nécessaires à l'opération, sans exiger
de répéter manuellement les données.

```python
# Ajouter une valeur à chaque élément
ventes2 = np.array([1000, 500, 4000, 1200])
resultat = ventes2 + 100
print(resultat)
```

```python
#Broadcasting sur une matrice
ventes3 = np.array([[100, 4500, 2000], [3000, 500, 22000]])

supplement = np.array([300, 400, 100])
resultat1 = ventes3 + supplement
print(resultat1)
```

# 5. Indexation et slicing

L'indexation permet d'accéder à une valeur précise.
Le slicing permet de sélectionner une partie d'un tableau.
Les indices NumPy commencent à zéro.

Nous allons voir :
- `arr[i]` : sélectionner un élément ;
- `arr[i:j]` : sélectionner une tranche ;
- `arr[mask]` : filtrer avec un tableau de booléens ;
- l'indexation des tableaux à deux dimensions.

```python
#Sélectionner un élément

mts = np.array([3000, 8400, 4000, 2100, 5300])

#premier elmnt: indice 0
print(mts[0])
#3eme elemnt: indice 2
print(mts[2])
#dernier elmnt
print(mts[-1])

#Sélectionner une tranche

# Sélectionner les éléments d'indice 1 inclus à 4 exclu
print(mts[1:4])
#selectionner les 3 premiers elmnt
print(mts[:3])
#selectionner a partir de l'indice 2
print(mts[2:])
#selectionne 1 elmn sur 2
print(mts[::2])
```

```python
#Indexation d'une matrice
#Trois produits et leurs ventes sur deux jours
ventes4 = np.array([
    [1000, 1500], 
    [2000, 2500], 
    [3000, 3500]
])

#Deuxieme prduit, premiere jrnée
print(ventes4[1, 0])
#Toutes les lignes, première colonne
print(ventes4[:, 0])
#Deuxième ligne entière
print(ventes4[1, :])
```

# 6. Masques booléens et filtrage

Un masque booléen est un tableau contenant des valeurs `True` et `False`.
Il permet de sélectionner uniquement les éléments qui respectent une condition.

Cette technique est fondamentale en analyse de données :
elle permet d'isoler des ventes élevées, des valeurs négatives
ou des observations qui respectent plusieurs critères.

```python
# Filtrer les montants positifs

montants2 = np.array([1200, -500, 2500, 0, 4000, -100])

#masque positif 
masque = montants2 > 0
print("Masque :", masque)

montants_positfs = montants2[masque]
print("valeurs positives :", montants_positfs)
print(montants2[montants2 > 0])

#Combiner plusieurs conditions
selection = montants2[(montants2 >= 1000) & (montants2 <= 3000)]
print(selection)
```

# 7. Fonctions statistiques

NumPy fournit des fonctions pour résumer des données numériques.

- `mean()` : moyenne ;
- `std()` : écart-type ;
- `sum()` : somme ;
- `min()` et `max()` : valeurs minimale et maximale ;
- `argmin()` et `argmax()` : indices des valeurs minimale et maximale.

Ces fonctions permettent d'obtenir rapidement des indicateurs
sur des séries de données.

```python
stats = np.array([2000, 1000, 2000, 3000, 4000, 5000, 2000])
print("Somme :", np.sum(stats))
print("Moyenne :", np.mean(stats))
print("Écart-type :", np.std(stats))
print("Minimum :", np.min(stats))
print("Maximum :", np.max(stats))
print("======================================")

indice_min = np.argmin(stats)
indice_max = np.argmax(stats)
print("Indice du minimum :", indice_min)
print("Indice du maximum :", indice_max)
print("Montants Max :", stats[indice_min])
print("Montant maximum :", stats[indice_max])
```

```python
# stats d'une matrice
stats_matrice = np.array([
    [1200, 3200, 9500],
    [2100, 7800, 3200],
    [3400, 2100, 5000]
])
#somme des valeurs
print("Somme totale :", np.sum(stats_matrice))
#somme par colonne
print("Somme par jour :", np.sum(stats_matrice, axis=0))
#somme par ligne
print("Somme par produit :", np.sum(stats_matrice, axis=1))
```

# 8. Trier les valeurs et identifier les valeurs uniques

`np.sort()` renvoie une version triée d'un tableau.

`np.unique()` renvoie les valeurs distinctes, classées par ordre croissant.

Ces fonctions servent notamment à explorer des valeurs numériques
et à repérer les catégories distinctes dans des données encodées.

```python
# Trier un tableau
tab = np.array([3000, 1000, 5000, 2000, 1000])
print("Trié :", np.sort(tab))

tab1 = np.array([101, 102, 101, 103, 102, 101])
print("Uniques :", np.unique(tab1))
```

# 9. Gestion des types et valeurs spéciales

Les tableaux NumPy utilisent un type de données (`dtype`) pour
définir la manière dont les valeurs sont représentées.

Il est important de comprendre :
- les types entiers et flottants ;
- la conversion des types ;
- les valeurs manquantes représentées par `np.nan` ;
- les valeurs infinies représentées par `np.inf`.

Une mauvaise gestion des types peut provoquer des erreurs
de calcul ou des résultats inattendus.

```python
donnee = np.array([2, 5, 8])
print(donnee.dtype)

prx = np.array([1500.5, 2000.75, 3500.0])
print(prx)

#Convertir un type
prx_convert = prx.astype(int)
print(prx_convert)
```

```python
#np.nan et les valeurs manquantes
mesures = np.array([10.0, 20.0, np.nan, 40.0])
print(mesures)

# la moy devient nan si le tab contient nan
print("Moyenne standard :", np.mean(mesures))
# la moy en ignorant Nan
print("Moyenne sans Nan :", np.nanmean(mesures))
#identifier les val manquantes
print("Masque NaN :", np.isnan(mesures))
```

```python
# Tableau contenant une valeur infinie positive
valeurs = np.array([10.0, 20.0, np.inf, -np.inf])

print(valeurs)

# Identifier les valeurs infinies
print(np.isinf(valeurs))

# Identifier les valeurs finies
print(np.isfinite(valeurs))
```

```python
# Mesures pouvant contenir des valeurs manquantes ou infinies
mesures1 = np.array([10.0, np.nan, 25.0, np.inf, 30.0])

# Identifier les valeurs utilisables dans un calcul classique
masque_fini = np.isfinite(mesures1)

print("Valeurs finies :", mesures1[masque_fini])
```
