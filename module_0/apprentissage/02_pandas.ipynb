# 1. Introduction et création d'un DataFrame

Un DataFrame est une structure tabulaire composée de lignes et de colonnes.
Chaque colonne peut avoir son propre type de données.

Pandas est particulièrement utile pour explorer, nettoyer et transformer
les données avant leur analyse ou leur chargement dans un système cible.

```python
# Importer Pandas et NumPy
import pandas as pd
import numpy as np

# Créer un DataFrame de commandes fictives
df = pd.DataFrame({
    "commande_id": [101, 102, 103, 104, 105],
    "ville": ["Dakar", "Thiès", "Dakar", "Saint-Louis", "Thiès"],
    "produit": ["Riz", "Huile", "Sucre", "Riz", "Huile"],
    "quantite": [2, 1, 5, 3, 2],
    "montant": [5000, 2500, 1500, 7500, 5000]
})

# Afficher le tableau
df
```

```python
# Afficher les premières lignes
print(df.head(3))
# Nombre de lignes et de colonnes
print("Dimensions :", df.shape)
# Noms des colonnes
print("Colonnes :", df.columns.tolist())
# Types des colonnes et valeurs non nulles
df.info()
# Statistiques descriptives des colonnes numériques
df.describe()

#shape renvoie (nombre_de_lignes, nombre_de_colonnes). info() sert à contrôler la structure, 
#tandis que describe() donne des statistiques numériques
```

# 2. Sélection et accès aux données

- `df["colonne"]` : sélectionner une colonne.
- `df[["col1", "col2"]]` : sélectionner plusieurs colonnes.
- `loc` : sélectionner par étiquettes et conditions.
- `iloc` : sélectionner par positions numériques.

```python
# Sélectionner une seule colonne : résultat de type Series
print(df["ville"])

# Sélectionner plusieurs colonnes : résultat de type DataFrame
print(df[["ville", "montant"]])

# Sélectionner les lignes dont l'indice va de 0 à 2 inclus
print(df.loc[0:2, ["produit", "montant"]])

# Sélectionner les lignes en positions 0 et 1, toutes les colonnes
print(df.iloc[0:2])
```

# 3. Filtrage, tri et création de colonnes

Le filtrage permet de ne conserver que les lignes répondant à une condition.
Le tri organise les résultats selon une ou plusieurs colonnes.
Une nouvelle colonne peut être calculée à partir des colonnes existantes.

```python
# Commandes dont le montant dépasse 4000 FCFA
print(df[df["montant"] > 4000])
#Commande de Dakar
print(df[df["ville"]=='Dakar']) 
# Combiner deux conditions : Dakar ET montant supérieur à 2000
print(df[(df["ville"] == "Dakar") & (df["montant"] > 2000)])
#trie par mt
print(df.sort_values("montant", ascending=False))
```

# 4. Nettoyage des données

Les données réelles contiennent souvent des valeurs manquantes,
des lignes dupliquées ou des colonnes dont le type est incorrect.

Le nettoyage consiste à identifier les problèmes, choisir une règle
adaptée et vérifier le résultat. Il ne faut pas supprimer des données
sans comprendre pourquoi elles sont problématiques.

```python
# Créer un jeu de données volontairement imparfait
df_sale = pd.DataFrame({
    "commande_id": [1, 2, 2, 3, 4],
    "ville": ["Dakar", "Thiès", "Thiès", None, "Dakar"],
    "montant": ["5000", "2500", "2500", None, "3500"]
})

# Compter les valeurs manquantes par colonne
print(df_sale.isna().sum())

# Compter les lignes dupliquées
print("Doublons :", df_sale.duplicated().sum())

# Convertir les montants en nombres ; les valeurs invalides deviennent NaN
df_sale["montant"] = pd.to_numeric(df_sale["montant"], errors="coerce")

# Supprimer les lignes entièrement dupliquées
df_sale = df_sale.drop_duplicates()
# Afficher le résultat nettoyé
print(df_sale)
print(df_sale.dtypes)
```

```python
# Travailler sur une copie pour préserver le DataFrame original
df_nettoye = df_sale.copy()
# Remplacer les villes manquantes par une catégorie explicite
df_nettoye["ville"] = df_nettoye["ville"].fillna("Inconnue")
# Supprimer les lignes sans montant, uniquement si ce choix est adapté au besoin
df_sans_montant_manquant = df_nettoye.dropna(subset=["montant"])
print(df_sans_montant_manquant)
```

# 5. Agrégation et analyse par groupe

`groupby()` permet de regrouper les lignes selon une ou plusieurs colonnes,
puis de calculer des indicateurs pour chaque groupe.

C'est une opération fondamentale pour calculer le chiffre d'affaires
par ville, les quantités vendues par produit ou le panier moyen.

```python
# CA par ville
ca_ville = df.groupby("ville")["montant"].sum()
print(ca_ville)

# qt totale par prdt 
qt_prdt = df.groupby("produit")["quantite"].sum()
print(qt_prdt)

# plusieurs indicateurs par ville
indicateurs = df.groupby("ville").agg(
    c_a = ("montant", "sum"),
    montant_moy = ("montant", "sum"),
    nbre_commandes = ("commande_id", "count")
)
print(indicateurs)
```

# 6. Jointures entre DataFrames

Les données sont souvent réparties dans plusieurs tables.
`merge()` permet de les relier grâce à une clé commune, comme un identifiant.

Les jointures servent notamment à associer les commandes aux clients,
ou les ventes aux informations produits.

```python
# Table contenant les informations clients
clients = pd.DataFrame({
    "client_id": [1, 2, 3],
    "nom_client": ["Awa", "Moussa", "Fatou"],
    "ville": ["Dakar", "Thiès", "Saint-Louis"]
})

# Table des commandes
commandes = pd.DataFrame({
    "commande_id": [101, 102, 103, 104],
    "client_id": [1, 2, 1, 3],
    "montant": [5000, 2500, 3000, 7000]
})

# Relier chaque commande aux informations du client
resultat = commandes.merge(
    clients,
    on="client_id",
    how="left"
)

print(resultat)
```

# 7. Dates et séries temporelles

Les dates importées depuis un fichier sont parfois des chaînes de caractères.
`pd.to_datetime()` permet de les convertir en valeurs temporelles,
pour ensuite extraire l'année, le mois ou le jour et analyser leur évolution.

```python
# Créer des commandes avec des dates stockées sous forme de texte
df_dates = pd.DataFrame({
    "commande_id": [1, 2, 3],
    "date_commande": ["2026-01-15", "2026-02-20", "2026-03-05"],
    "montant": [5000, 3000, 7000]
})

# Convertir la colonne en dates
df_dates["date_commande"] = pd.to_datetime(
    df_dates["date_commande"],
    errors="coerce"
)

# Extraire l'année et le mois
df_dates["annee"] = df_dates["date_commande"].dt.year
df_dates["mois"] = df_dates["date_commande"].dt.month

# Calculer le chiffre d'affaires par mois
print(df_dates.groupby("mois")["montant"].sum())
```

# 8. Importer et exporter des données

Pandas permet de lire et d'écrire plusieurs formats de données courants.

- CSV : format tabulaire textuel, simple et largement utilisé.
- JSON : format structuré couramment utilisé par les API.
- Parquet : format colonne adapté aux données analytiques.

Le format choisi dépend de la source, de l'usage et des contraintes du projet.

```python
# Exporter le DataFrame vers un fichier CSV
df.to_csv("commandes.csv", index=False)

# Relire le fichier CSV
df_csv = pd.read_csv("commandes.csv")

# Exporter en JSON orienté enregistrements
df.to_json("commandes.json", orient="records", force_ascii=False, indent=2)

# Relire le fichier JSON
df_json = pd.read_json("commandes.json")

# Exporter en Parquet (nécessite pyarrow ou fastparquet)
df.to_parquet("commandes.parquet", index=False)

# Relire le fichier Parquet
df_parquet = pd.read_parquet("commandes.parquet")

print(df_csv.head())
print(df_json.head())
print(df_parquet.head())
```
