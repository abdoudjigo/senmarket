# 04 — Linux et Shell pour le Data Engineering

## Objectifs

- Naviguer dans le système de fichiers Linux.
- Créer, déplacer, copier et supprimer des fichiers.
- Lire et rechercher des informations dans des fichiers.
- Manipuler des données textuelles depuis le terminal.
- Utiliser les redirections et les pipes.
- Écrire des scripts Bash pour automatiser des tâches.

## 1. Terminal, Shell et commandes

- **Terminal** : interface permettant d'interagir avec le système.
- **Shell** : programme qui interprète les commandes saisies, par exemple Bash.
- **Commande** : instruction exécutée par le Shell.

En Data Engineering, Linux sert notamment à gérer les fichiers de données,
exécuter des pipelines, consulter des logs et automatiser des traitements.


## 2. Lire et rechercher des fichiers

Les commandes suivantes permettent de consulter le contenu des fichiers
et de rechercher des informations sans ouvrir un éditeur de texte.

- `cat` : affiche tout le contenu d'un fichier.
- `head` : affiche les premières lignes.
- `tail` : affiche les dernières lignes.
- `grep` : recherche un texte précis.
- `find` : recherche des fichiers et des dossiers.



## 3. Ce que je retiens

- `cat` permet de lire un fichier.
- `head` permet d'afficher le début d'un fichier.
- `grep` permet de rechercher une valeur.
- `wc -l` permet de compter les lignes.

Ces commandes sont utiles pour vérifier rapidement des fichiers CSV
avant de les intégrer dans un pipeline de données.



## 4. Redirections et pipes

### 4.1. Les redirections

Les redirections permettent d'envoyer le résultat d'une commande vers un fichier.

- `>` : écrit le résultat dans un fichier en remplaçant son contenu.
- `>>` : ajoute le résultat à la fin d'un fichier.
- `<` : utilise un fichier comme entrée d'une commande.

**Exemple :** `ls > fichiers.txt` enregistre la liste des fichiers dans `fichiers.txt`.

### 4.2. Les pipes

Le symbole `|` transmet le résultat d'une commande à une autre commande.

**Exemple :** `cat ventes.csv | grep Dakar` affiche uniquement les lignes contenant « Dakar ».

En Data Engineering, ces mécanismes permettent d'automatiser des traitements
et d'examiner rapidement les fichiers de données.



### 4.3. Synthèse

| Symbole | Rôle |
|---|---|
| `>` | Écrire ou remplacer le contenu d'un fichier |
| `>>` | Ajouter du contenu à un fichier |
| `<` | Utiliser un fichier comme entrée |
| `|` | Transmettre le résultat à une autre commande |

Ces outils sont utiles pour examiner les données, conserver des résultats
et construire des traitements automatisés en ligne de commande.



## 5. Manipuler les données textuelles

Linux fournit des commandes pour analyser rapidement les fichiers texte et CSV.

- `wc -l` : compte les lignes.
- `sort` : trie les lignes.
- `uniq` : identifie les lignes consécutives identiques.
- `cut` : extrait des colonnes ou des champs.
- `awk` : analyse les lignes et sélectionne des champs.

Ces commandes permettent de réaliser des contrôles rapides avant de traiter
les données avec Python, Pandas ou un outil ETL.



## 6. Automatiser avec Bash

Un script Bash est un fichier contenant plusieurs commandes Linux exécutées
dans l'ordre.

Il permet d'automatiser des tâches répétitives : vérifier des fichiers,
créer des dossiers, lancer des traitements ou produire des logs.

Notions importantes :

- Variables : stocker des valeurs.
- Conditions : exécuter des instructions selon un résultat.
- Permissions : autoriser l'exécution d'un script.
- Logs : conserver des informations sur l'exécution.



## 7. Récapitulatif

| Besoin | Commande |
|---|---|
| Connaître le dossier courant | `pwd` |
| Lister les fichiers | `ls` |
| Lire un fichier | `cat` |
| Rechercher du texte | `grep` |
| Compter les lignes | `wc -l` |
| Extraire un champ | `cut` |
| Traiter des champs | `awk` |
| Rediriger un résultat | `>` et `>>` |
| Enchaîner des commandes | `|` |
| Automatiser des commandes | Script Bash |
| Modifier les permissions | `chmod` |

Linux et Bash sont essentiels pour gérer les fichiers, contrôler les données
et automatiser des traitements en Data Engineering.
