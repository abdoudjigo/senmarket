## 1. Les trois formats essentiels

### CSV — Comma-Separated Values

Format textuel qui stocke les données sous forme de lignes et de colonnes.
Les valeurs sont généralement séparées par des virgules ou des points-virgules.

**Avantages :** simple, lisible, compatible avec de nombreux outils.

**Limites :** les types de données ne sont pas conservés de manière fiable
et les fichiers peuvent être volumineux.

### JSON — JavaScript Object Notation

Format textuel qui représente les données sous forme d'objets et de listes.

**Avantages :** adapté aux API et aux données semi-structurées.

**Limites :** peut être verbeux et nécessite parfois une étape de normalisation
pour obtenir un tableau exploitable.

### Parquet

Format binaire en colonnes conçu pour le stockage et l'analyse de données.

**Avantages :** conserve les types de données, permet une lecture sélective
des colonnes et offre généralement une bonne compression.

**Limites :** moins lisible directement par un humain et nécessite un outil compatible.


```python

import pandas as pd
from pathlib import Path

# Créer un dossier pour stocker les fichiers
dossier = Path("data_formats")
dossier.mkdir(exist_ok=True)

# Créer un DataFrame de ventes
df = pd.DataFrame({
    "commande_id": [1, 2, 3, 4],
    "ville": ["Dakar", "Thiès", "Dakar", "Saint-Louis"],
    "produit": ["Riz", "Mil", "Huile", "Riz"],
    "quantite": [3, 5, 2, 4],
    "montant": [15000, 8000, 5000, 20000]
})

# Afficher les données
display(df)
```

```python

# CSV : données tabulaires en texte
df.to_csv(dossier / "ventes.csv", index=False)

# JSON : données structurées en objets
df.to_json(
    dossier / "ventes.json",
    orient="records",
    indent=2,
    force_ascii=False
)

# Parquet : données stockées en colonnes
# Nécessite pyarrow ou fastparquet
df.to_parquet(dossier / "ventes.parquet", index=False)

print("Les trois fichiers ont été créés.")
```

```python

# Lire le CSV
df_csv = pd.read_csv(dossier / "ventes.csv")

# Lire le JSON
df_json = pd.read_json(dossier / "ventes.json")

# Lire le Parquet
df_parquet = pd.read_parquet(dossier / "ventes.parquet")

# Vérifier le résultat
print("CSV :")
display(df_csv)

print("JSON :")
display(df_json)

print("Parquet :")
display(df_parquet)
```

```python

# Vérifier que les données relues sont identiques
print("CSV identique :", df.equals(df_csv))
print("JSON identique :", df.equals(df_json))
print("Parquet identique :", df.equals(df_parquet))
# Vérifier les types de données
print("\nTypes dans le DataFrame original :")
print(df.dtypes)
print("\nTypes après lecture du CSV :")
print(df_csv.dtypes)
print("\nTypes après lecture du Parquet :")
print(df_parquet.dtypes)
```


## 2. Comparaison des formats

| Critère | CSV | JSON | Parquet |
|---|---|---|---|
| Lisibilité humaine | Excellente | Bonne | Faible |
| Données tabulaires | Très adapté | Possible | Très adapté |
| Données imbriquées | Peu adapté | Très adapté | Possible |
| Conservation des types | Limitée | Dépend de la structure | Bonne |
| Analyse à grande échelle | Moins performant | Variable | Souvent performant |
| Cas courant | Échange de fichiers | API et événements | Data Lake et analytique |

## 3. À retenir

- CSV : échange simple de données tabulaires.
- JSON : échanges avec les API et données semi-structurées.
- Parquet : stockage analytique efficace et pipelines de données.

Le choix dépend du système source, du volume, de la structure et de l'utilisation
future des données.


```python

# Afficher le nom et la taille de chaque fichier
for fichier in sorted(dossier.iterdir()):
    if fichier.is_file():
        taille_ko = fichier.stat().st_size / 1024
        print(f"{fichier.name} : {taille_ko:.2f} Ko")
```
